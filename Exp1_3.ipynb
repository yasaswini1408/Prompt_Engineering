```python
import os
from google.colab import userdata
from langchain_core.messages import HumanMessage
from langchain.chat_models import init_chat_model
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_groq import ChatGroq
```

```python
os.environ.update({
    "GOOGLE_API_KEY": userdata.get('GOOGLE_API_KEY'),
    "GROQ_API_KEY": userdata.get('GROQ_API_KEY'),
    "OPENROUTER_API_KEY": userdata.get('OPENROUTER_API_KEY'),
    "HF_TOKEN": userdata.get('HF_TOKEN')
    })
```

```python
models = {
    # Explicit initialization via provider-specific classes
    "Google": ChatGoogleGenerativeAI(
        model="gemini-3.1-flash-lite",
        temperature=1.0
        ),
    "Groq": ChatGroq(
        model_name="openai/gpt-oss-120b",
        temperature=1.0
        ),
    # Universal initialization via LangChain's init_chat_model
    "OpenRouter": init_chat_model(
        "nvidia/nemotron-3-super-120b-a12b:free",
        model_provider="openrouter",  temperature=1.0
        ),
    "HuggingFace": init_chat_model(
        "microsoft/Phi-3-mini-4k-instruct",
        model_provider="huggingface",
        temperature=1.0
    )
}
```

```python
msg = input("\nEnter your prompt: ")
for llm_name, llm in models.items():
  print(f"\n--- {llm_name} ---")
# Invoke the model and display the response
response = llm.invoke(msg)
print(f"{response}")
```

```python
import os
from google.colab import userdata
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
```

```python
os.environ["GOOGLE_API_KEY"] = userdata.get('GOOGLE_API_KEY')
llm = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite", temperature=0.3)
```

```python
print("--- Enter Prompt Anatomy Details ---")
data = {  k: input(f"Enter {k.capitalize()}: ")
for k in [
    "role", "task", "topic", "context",  "tone", "format", "constraints", "goal"
    ]
}
```

```python
registry = {
    "v1_naive": PromptTemplate.from_template(
        "Write about {topic}."
        ),
    "v2_intermediate": PromptTemplate.from_template(
        "Act as a {role}. {task} about {topic} in a {tone} tone."
        ),
    "v3_enhanced": PromptTemplate.from_template(
        "Role: {role}\nTask: {task} on Topic: {topic}\nContext: {context}\n"
        "Tone: {tone}\nFormat: {format}\nConstraints: {constraints}\nGoal: {goal}"
        )
      }
```

```python
for ver, temp in registry.items():
  print(f"\n--- EXECUTING {ver.upper()} ---")
  # Dynamically inject only the keys required by the specific template
  formatted = temp.format(
      **{k: v for k, v in data.items() if k in temp.input_variables}
    )
  print(f"Result:\n{llm.invoke(formatted).content[0]["text"]}\n")
```

```python
import os
from google.colab import userdata
from langchain_groq import ChatGroq
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
```

```python
os.environ["GROQ_API_KEY"] = userdata.get('GROQ_API_KEY')
llm = ChatGroq(model="openai/gpt-oss-120b", temperature=0.1)
```

```python
sys_msg = SystemMessage(
    content="Role: Expert NLP Parser. Task: Perform Part-of-Speech (POS) "
    "tagging. Output STRICTLY in this custom format: "  "Word => [TAG] || Word => [TAG]"
    )
query = HumanMessage(content=input("Enter sentence to parse: "))
```

```python
ex_h1 = HumanMessage(content="I write code.")
ex_a1 = AIMessage(  content="I => [PRON] || write => [VERB] || code => [NOUN] || . => [PUNCT]"  )
ex_h2 = HumanMessage(content="Red birds fly fast.")
ex_a2 = AIMessage(  content="Red => [ADJ] || birds => [NOUN] || fly => [VERB] || "  "fast => [ADV] || . => [PUNCT]"  )
ex_h3 = HumanMessage(content="Wow, it works perfectly!")
ex_a3 = AIMessage(  content="Wow => [INTJ] || , => [PUNCT] || it => [PRON] || "  "works => [VERB] || perfectly => [ADV] || ! => [PUNCT]"  )
```

```python
scenarios = {
    "v1_zero_shot": [sys_msg, query],
    "v2_one_shot": [sys_msg, ex_h1, ex_a1, query],
    "v3_few_shot": [sys_msg, ex_h1, ex_a1, ex_h2, ex_a2, ex_h3, ex_a3, query]
    }
```

```python
for name, messages in scenarios.items():
   print(f"\n--- EXECUTING {name.upper()} ---")
   print(f"Result:\n{llm.invoke(messages).content}")
```
