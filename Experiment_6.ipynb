```python
# ===================================================================== # Step 1: Install Required Libraries # =====================================================================
!pip install langchain-google-genai langchain_core
# ===================================================================== # Step 2: Import Necessary Libraries # =====================================================================
import os
from langchain_core.prompts import ChatPromptTemplate,FewShotChatMessagePromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
from google.colab import userdata
# ===================================================================== # Step 3: Initialization & Setup # =====================================================================
os.environ["GOOGLE_API_KEY"] = userdata.get('GOOGLE_API_KEY')
llm = ChatGoogleGenerativeAI(     model="gemini-3.1-flash-lite",
                             temperature=0.0,
                                  thinking_level="high",
                                  include_thoughts=True )
# ===================================================================== # Step 4: Define Two Simple Few-Shot Math Examples # =====================================================================
examples = [     {
    "input": "Sam has 5 boxes of pencils. Each box has 2 pencils. He loses 3 pencils. How many does he have left?",
    "output": (
        "1. TASK DECOMPOSITION:\n"
        "   - Total boxes = 5\n"
        "   - Pencils per box = 2\n"
        "   - Lost pencils = 3\n\n"
        "2. CHRONOLOGICAL CHAIN-OF-THOUGHT CALCULATION:\n"
        "   - Step A: Find total starting pencils by multiplying boxes by pencils per box: 5 boxes * 2 pencils = 10 pencils.\n"
        "   - Step B: Subtract the lost pencils from the total starting amount: 10 pencils - 3 lost = 7 pencils.\n\n"
        "FINAL ANSWER VERDICT:\n"
        "Sam has 7 pencils left."
        )
    },
         {
             "input": "Tina buys 3 candies for $4 each. She pays with a $20 bill. How much change does she get?",
             "output": (
                 "1. TASK DECOMPOSITION:\n"
                 "   - Number of candies = 3\n"
                 "   - Cost per candy = $4\n"
                 "   - Cash given = $20\n\n"
                 "2. CHRONOLOGICAL CHAIN-OF-THOUGHT CALCULATION:\n"
                 "   - Step A: Calculate total cost of candies by multiplying quantity by price: 3 candies * $4 = $12 total cost.\n"
                 "   - Step B: Calculate change by subtracting total cost from cash given: $20 cash - $12 cost = $8 change.\n\n"
                 "FINAL ANSWER VERDICT:\n"
                 "Tina gets $8 change."
                 )
            } ]
                  # ===================================================================== # Step 5: Assemble the Few-Shot Prompt Template # =====================================================================
example_prompt = ChatPromptTemplate.from_messages([
                ("human", "{input}"),
                ("ai", "{output}") ])
few_shot_prompt = FewShotChatMessagePromptTemplate(
                      example_prompt=example_prompt,
                      examples=examples )
final_prompt = ChatPromptTemplate.from_messages([
                      ("system", (
                          "You are an expert Math Tutor AI. When given a word problem, you must never jump straight to the answer. "
                          "You must first decompose the problem to list the known variables, and then show your step-by-step "
                          "algebraic Chain-of-Thought calculations before providing the final answer verdict."
                          )),
                      few_shot_prompt,
                       ("human", "{input}") ])
                  # Assemble the pure LCEL chain
math_chain = final_prompt | llm
                  # ===================================================================== # Step 6: Runtime Execution # =====================================================================
print("--- Math Reasoning Workflow (CoT & Decomposition) ---")
                  # A simple runtime problem for the model to process using the learned patterns
runtime_problem = input("Enter the math problem: ")
print(f"\nUser Math Problem: {runtime_problem}")
print("\n--- Executing Chain-of-Thought & Task Decomposition Pipeline ---")
response = math_chain.invoke({"input": runtime_problem})
                  # ===================================================================== # Step 7: Output Results (Thinking Process & Final Answer) # =====================================================================
print("\n--- Generated LLM Reasoning or Thinking ---")
print(response.content[0]["thinking"])
print("\n--- Generated LLM output ---")
print(response.content[1]["text"])
```
